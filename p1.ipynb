```python
import pandas as p
import matplotlib.pyplot as mp
import seaborn as sn

with open('etudiants.csv') as f:
    table_etud=p.read_csv(f)
    print(table_etud)
    print( table_etud.isna().sum())
    print( table_etud.duplicated())
    table_etud.fillna({"id":4},inplace=True)
    table_etud.dropna(subset=["filière","id_etudiant"],inplace=True)
    table_etud.drop_duplicates(inplace=True)

with open("livre.csv") as m:
    table_liv=p.read_csv(m)
    table_liv.dropna(inplace=True)
    table_liv.fillna({"categorie":"Educatif"},inplace=True)

with open("emprunts.csv") as n:
    table_emp=p.read_csv(n)
    table_emp.fillna({"durée":4},inplace=True)
# table_emp=table_emp.set_index("id_etudiant")
# table_edu_emp=table_etud.join(table_emp,on='id_etudiant')
table_edu_emp=p.merge(table_etud,table_emp,on=['id_etudiant'])
table_emp_liv=p.merge(table_edu_emp,table_liv,on='id_livre')
print(table_edu_emp)
print(table_emp_liv)
table_liv_emp=table_emp.set_index("id_livre")
table_liv_emp1=table_liv.join(table_liv_emp,on="id_livre")
# cu=table_edu_emp.groupby("filière")["id_etudiant"].value_counts()
cu=table_edu_emp["filière"].value_counts()
bu=table_liv_emp1.groupby("categorie")["durée"].mean()
bu1=table_liv_emp1.pivot_table(index='categorie',values='durée',aggfunc='mean')
print(bu) 
# sn.barplot(data=cu,edgecolor='red')
cu.plot(kind='bar')
table_emp.corr(numeric_only=True)
mp.show()
mp.show()
sn.scatterplot(data=table_emp,x='id_livre',y='durée')
mp.scatter(x=table_liv_emp1['categorie'],y=table_liv_emp1['durée'])
mp.show()
table_etud.to_csv('etudiants.csv',index=False)
table_liv.to_csv('livre.csv',index=False)
table_emp.to_csv('emprunts.csv',index=False)
```
